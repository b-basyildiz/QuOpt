```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_fidelity(
    csv_path: str,
    graph_type: str = "CNOT",
    label: str | None = None,
    ax: plt.Axes | None = None,
    save_path: str | None = None,
) -> plt.Axes:
    plt.rcParams["text.usetex"] = False
    plt.rcParams.update({
        "font.family": "DejaVu Sans",
        "mathtext.fontset": "cm",
    })
    df = pd.read_csv(csv_path, header=None, names=["F", "T_over_Tmin"])
    df = (
        df.groupby("T_over_Tmin", as_index=False)["F"]
        .max()
        .sort_values("T_over_Tmin")
    )
    if ax is None:
        ax = plt.gca()
    ax.plot(df["T_over_Tmin"], df["F"], marker="o", label=label)
    ax.set_xlabel(r"$T/T_{min}^{ %s}$" % graph_type, fontsize=16)
    ax.set_ylabel(r"$F$", fontsize=16)
    ax.grid(True)
    handles, labels = ax.get_legend_handles_labels()
    if labels:
        ax.legend(fontsize=13)
    if save_path:
        ax.get_figure().tight_layout()
        ax.get_figure().savefig(save_path)
    return ax
```

```python
def capacitiveCoup_EC(level):
    """
    Excitation-conserving part of the capacitive coupling:
    H = kron(a†, a) + kron(a, a†)
    Keeps only terms where one qudit loses an excitation and the other gains one.
    Equivalent to 'AshhabHopp' in genCouplMat.
    """
    a = np.zeros((level, level))
    for i in range(level - 1):
        a[i, i+1] = np.sqrt(i + 1)   # annihilation operator
    adag = a.T                         # creation operator
    return kron(adag, a) + kron(a, adag)

H0_EC = capacitiveCoup_EC(3)
printMat(H0_EC)
```

```python
# Reconstruct the qutrit CZ unitary from weights and compute fidelity.
weights_path = "./coupling_data/capacitiveCoup/Weights/qutrit_CZ_capacitiveCoup_M40_SGD_SRK2_g1.0_maxT2.0_maxD40_it5000_Weights/Weights_t1.0.csv"
weights = pd.read_csv(weights_path, header=None).to_numpy()
print("Loaded weights:", weights.shape, "from", weights_path)

level = 3
N = 2
M = weights.shape[0]
maxDriveStrength = 40
t_ratio = 1.0
tmin = np.pi / 4
gate_time = t_ratio * tmin

H0 = torch.tensor(genCouplMat("capacitiveCoup", level), dtype=torch.cdouble)

id_level = np.eye(level)
drives = []
for l in range(1, level):
    drives.append(genDrive(level, l, "x"))
    drives.append(genDrive(level, l, "y"))


def sum_pauli(coef, gate):
    total_pauli = torch.zeros((level ** N, level ** N), dtype=torch.cdouble)
    for i in range(N):
        pauli_temp = 1
        for j in range(i):
            pauli_temp = np.kron(pauli_temp, id_level)
        pauli_temp = np.kron(pauli_temp, gate)
        for j in range(i + 1, N):
            pauli_temp = np.kron(pauli_temp, id_level)
        total_pauli = total_pauli + maxDriveStrength * torch.cos(coef[i]) * torch.tensor(pauli_temp, dtype=torch.cdouble)
    return total_pauli

U = torch.eye(level ** N, dtype=torch.cdouble)
for m in range(M):
    pulse_coef = torch.tensor(weights[m], dtype=torch.double)
    H1 = torch.zeros((level ** N, level ** N), dtype=torch.cdouble)
    for i, d in enumerate(drives):
        H1 = H1 + sum_pauli(pulse_coef[i * N:(i + 1) * N], d)
    H = H0 + H1
    U = torch.matmul(torch.matrix_exp(-1j * H * (gate_time / M)), U)

input_gate = torch.tensor(gateGen("CZ", level), dtype=torch.cdouble)

SU = genTwoQuditBasis(level, level, torch.cdouble)

fid = 0
for U_basis in SU:
    eps_U = torch.matmul(torch.matmul(U, U_basis), U.conj().T)
    target_U = torch.matmul(torch.matmul(input_gate, U_basis.conj().T), input_gate.conj().T)
    fid = fid + torch.trace(torch.matmul(target_U, eps_U))

d2 = level ** N
fidelity = abs((fid + d2 ** 2) / ((d2 ** 2) * (d2 + 1)))
print(f"Reconstructed qutrit CZ fidelity = {fidelity:.10f}")
```
